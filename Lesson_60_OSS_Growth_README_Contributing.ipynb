## The Concept: OSS Growth Is a Funnel, Not a Vibe

Treat your repo like a product with an acquisition funnel. Every stage has a specific job, and every stage has drop-off:

```
Discovery  →  First Impression  →  Activation   →  Contribution   →  Retention
(GitHub      (README, 30 sec)     (pip install     (good-first-      (repeat
 search,                           → works in       issue → PR       contributors,
 HN, Reddit,                       under 5 min)      → merged)        maintainer
 awesome-list)                                                        trust)
```

**Discovery** — people find the repo via GitHub topics, an awesome-list PR, a Show HN post, or a search engine. You don't control this directly, but you control whether the repo is *findable* (topics, description, keywords in README).

**First Impression** — the README is read for ~30 seconds before a stranger decides "install" or "back button." Badges, a one-line hook, and a copy-pasteable quickstart do the persuading here.

**Activation** — the gap between "I installed it" and "I got a result." If `pip install paper-distiller` doesn't produce a working digest within a few minutes, you lose the user. This is where L58's CLI polish and L59's clean API errors pay off.

**Contribution** — a stranger becomes a contributor only if there's an obvious, low-risk entry point: a `good-first-issue` label, a CONTRIBUTING.md that removes ambiguity, and a PR template that sets expectations.

**Retention** — the rare, valuable outcome: someone contributes *again*. This is earned by fast, respectful review — not by more marketing.

Today we build the artifacts for the first three stages properly, and the scaffolding for the last two.

## The README: Anatomy of a 30-Second Sell

A README that converts has a strict order of operations. Skip the order and you lose readers before they reach the part that would have convinced them.

1. **Name + one-line hook** — what it does, who it's for, in one sentence.
2. **Badges** — social proof and health signals, scannable in under a second.
3. **30-second quickstart** — copy-paste 3 commands, get a real result. No prerequisites explanation yet.
4. **Why this and not X** — a comparison table against the obvious alternatives (manual reading, other summarizers).
5. **Architecture** — one diagram, for the reader who's now curious how it works.
6. **Full usage** — CLI reference, API reference, links to docs.
7. **Contributing** — one paragraph + link to CONTRIBUTING.md, with the good-first-issue call to action.
8. **License**.

Notice what's *not* near the top: installation troubleshooting, the full architecture explanation, the roadmap. Those are for readers who already decided to stay — put them below the fold.

## Badges: What Each One Actually Communicates

Badges aren't decoration — each one answers a specific trust question a stranger has before installing:

| Badge | Question it answers | Wired to |
|---|---|---|
| CI status | "Does this even work right now?" | `.github/workflows/ci.yml` badge URL (auto-generated by GitHub Actions) |
| PyPI version | "Is this actually published, and what version?" | `https://img.shields.io/pypi/v/<pkg>.svg` — reads live from PyPI |
| Python versions | "Will this work with my Python?" | `https://img.shields.io/pypi/pyversions/<pkg>.svg` |
| License | "Can I use this at work?" | Static badge, must match `LICENSE` file exactly |
| Coverage | "Is this tested, or a demo?" | Codecov/Coveralls badge — requires CI upload step (L54's `--cov-report`) |
| Downloads | "Is anyone else using this?" | pepy.tech — free, no config needed beyond the PyPI name |

**The failure mode to avoid:** a badge that lies. A green CI badge on a repo whose tests are actually broken (because the workflow doesn't fail on test failure) is worse than no badge — it's a trust violation the moment someone checks. Every badge here must be backed by a real, working check.


````
def badge_row(pkg_name: str, gh_org: str, gh_repo: str) -> str:
    """Generate a verified badge row — call this whenever you rename the package or repo."""
    base = f"https://github.com/{gh_org}/{gh_repo}"
    return "\n".join([
        f"[![CI]({base}/actions/workflows/ci.yml/badge.svg)]({base}/actions/workflows/ci.yml)",
        f"[![PyPI version](https://img.shields.io/pypi/v/{pkg_name}.svg)](https://pypi.org/project/{pkg_name}/)",
        f"[![Python versions](https://img.shields.io/pypi/pyversions/{pkg_name}.svg)](https://pypi.org/project/{pkg_name}/)",
        "[![License: MIT](https://img.shields.io/badge/License-MIT-yellow.svg)](https://opensource.org/licenses/MIT)",
        f"[![codecov](https://codecov.io/gh/{gh_org}/{gh_repo}/branch/main/graph/badge.svg)](https://codecov.io/gh/{gh_org}/{gh_repo})",
        f"[![Downloads](https://static.pepy.tech/badge/{pkg_name})](https://pepy.tech/project/{pkg_name})",
    ])

print(badge_row("paper-distiller", "gouravkhanijoe", "paper-distiller"))

# 💡 EXPERIMENT: regenerate this after you actually create the GitHub repo —
# the CI badge URL 404s silently (renders as "no status") until a workflow has run at least once.
````

## The Contributor Funnel

````
CONTRIBUTING_MD = r"""# Contributing to paper-distiller

Thanks for considering a contribution! This project is built to be approachable for a first-time contributor.

## Quick setup

```bash
git clone https://github.com/gouravkhanijoe/paper-distiller
cd paper-distiller
pip install -e ".[dev]"
pytest                 # should be green before you change anything
```

## Finding something to work on

- Browse issues labeled [`good-first-issue`](https://github.com/gouravkhanijoe/paper-distiller/labels/good-first-issue) — scoped to a single file, no design decisions required.
- [`help-wanted`](https://github.com/gouravkhanijoe/paper-distiller/labels/help-wanted) — larger, maintainer-supported.
- Have your own idea? Open an issue first using the Feature Request template — avoids wasted work if it doesn't fit the roadmap.

## Making a change

1. Fork, branch off `main`: `git checkout -b fix/short-description`
2. Write the change + a test. PRs without tests for new behavior will be asked to add one.
3. Run `ruff check . && mypy paper_distiller && pytest --cov=paper_distiller`
4. Use [Conventional Commits](https://www.conventionalcommits.org/) for your commit message: `fix:`, `feat:`, `docs:`, `test:`.
5. Open the PR — the template will prompt you for what changed and why.

## Review process

- A maintainer responds within 48 hours (usually faster).
- CI must be green (lint, type-check, tests, coverage ≥ 80%) before merge.
- We squash-merge, so commit history inside your branch can be messy — don't worry about it.

## Code of Conduct

This project follows the [Contributor Covenant](CODE_OF_CONDUCT.md). Be kind.
"""

contrib_path = PROJECT_ROOT / "CONTRIBUTING.md"
contrib_path.write_text(CONTRIBUTING_MD)
print(f"Wrote {contrib_path} ({len(CONTRIBUTING_MD)} chars)")
````

## Issue Templates: GitHub Issue Forms

Modern GitHub uses **YAML issue forms** (not the old markdown templates) — they render as a structured form in the browser, which forces the reporter to supply the fields you actually need (repro steps, version, arXiv ID that failed) instead of a blank textbox that gets filled with "it doesn't work."

Three files go in `.github/ISSUE_TEMPLATE/`:
- `bug_report.yml`
- `feature_request.yml`
- `config.yml` — disables the blank-issue option and adds contact links (e.g., a Discussions link for questions, keeping bug/feature signal clean)


````
BUG_REPORT_YML = r"""name: Bug Report
description: Something isn't working as expected
title: "[Bug]: "
labels: ["bug", "triage"]
body:
  - type: input
    id: arxiv-id
    attributes:
      label: arXiv ID (if applicable)
      placeholder: "1706.03762"
  - type: textarea
    id: what-happened
    attributes:
      label: What happened?
      description: A clear description of the bug, plus what you expected instead.
    validations:
      required: true
  - type: textarea
    id: repro
    attributes:
      label: Steps to reproduce
      placeholder: |
        1. paper-distiller distill 1706.03762
        2. ...
    validations:
      required: true
  - type: input
    id: version
    attributes:
      label: paper-distiller version
      placeholder: "pip show paper-distiller"
    validations:
      required: true
  - type: dropdown
    id: python-version
    attributes:
      label: Python version
      options: ["3.10", "3.11", "3.12", "Other (specify below)"]
    validations:
      required: true
"""

FEATURE_REQUEST_YML = r"""name: Feature Request
description: Suggest an idea for this project
title: "[Feature]: "
labels: ["enhancement"]
body:
  - type: textarea
    id: problem
    attributes:
      label: What problem does this solve?
      description: Describe the use case — not just the feature.
    validations:
      required: true
  - type: textarea
    id: proposal
    attributes:
      label: Proposed solution
    validations:
      required: false
  - type: checkboxes
    id: willing
    attributes:
      label: Contribution
      options:
        - label: I'd be willing to open a PR for this
"""

CONFIG_YML = r"""blank_issues_enabled: false
contact_links:
  - name: Questions & Discussion
    url: https://github.com/gouravkhanijoe/paper-distiller/discussions
    about: Ask usage questions here instead of opening an issue.
"""

PR_TEMPLATE_MD = r"""## What does this PR do?

## Why?

## How was this tested?

- [ ] Added/updated tests
- [ ] `pytest` passes locally
- [ ] `ruff check .` and `mypy paper_distiller` pass locally

## Related issue

Closes #
"""

(ISSUE_TEMPLATE_DIR / "bug_report.yml").write_text(BUG_REPORT_YML)
(ISSUE_TEMPLATE_DIR / "feature_request.yml").write_text(FEATURE_REQUEST_YML)
(ISSUE_TEMPLATE_DIR / "config.yml").write_text(CONFIG_YML)
(GITHUB_DIR / "PULL_REQUEST_TEMPLATE.md").write_text(PR_TEMPLATE_MD)

for f in [ISSUE_TEMPLATE_DIR / "bug_report.yml", ISSUE_TEMPLATE_DIR / "feature_request.yml",
          ISSUE_TEMPLATE_DIR / "config.yml", GITHUB_DIR / "PULL_REQUEST_TEMPLATE.md"]:
    print(f"Wrote {f}")
````

## Launch Strategy: Soft Launch, Then Hard Launch

Don't post to Hacker News on day one. The sequence that actually works:

1. **Soft launch (private)** — send the repo to 3-5 people who'll actually try it and won't be polite. Fix what breaks. This is where you discover the README lied about the quickstart working.
2. **Seed activity** — file 5-10 real `good-first-issue`s yourself from your own TODO list. A repo with zero open issues looks abandoned, not finished.
3. **Dogfood publicly** — use it for something real and link back (e.g., "I used paper-distiller to summarize my reading list, here's the output").
4. **Hard launch** — one well-timed post, not five simultaneous ones:
   - **Show HN** (`Show HN: paper-distiller — ...`) — Tuesday–Thursday, 8-10am PT, in the post explain *what you personally use it for*, not just what it does.
   - **r/MachineLearning** or **r/LocalLLaMA** — if genuinely relevant, follow the subreddit's self-promo rules exactly (many require a "Discussion" framing, not a pure link).
   - **Twitter/X thread** — lead with the *result* (a real digest screenshot), not the announcement.
5. **Awesome-list PRs** — search `awesome-llm-apps`, `awesome-arxiv-tools` style repos and open a one-line addition PR. Low effort, durable discovery channel (people browse these lists for months after your launch day is forgotten).

**The mistake to avoid:** launching before the activation funnel works. If `pip install paper-distiller && paper-distiller distill <id>` fails for a stranger during the HN launch window, that traffic is wasted — worse, it's negative signal (bad first impressions travel).


````
LAUNCH_CHECKLIST_MD = r"""# Launch Checklist — paper-distiller v1.0

## Before any public post
- [ ] `pip install paper-distiller` works on a clean machine/venv (test this literally, don't assume)
- [ ] README quickstart commands copy-pasted and verified end to end
- [ ] CI badge is green (not just present)
- [ ] 5+ issues labeled `good-first-issue`
- [ ] CONTRIBUTING.md reviewed by someone who isn't you
- [ ] License file present and matches badge

## Soft launch
- [ ] Sent to 3-5 people outside the project for unfiltered feedback
- [ ] Fixed at least the top 2 friction points they hit

## Hard launch (pick ONE per week, don't stack)
- [ ] Show HN post drafted — leads with personal use case
- [ ] Relevant subreddit post drafted, self-promo rules checked
- [ ] Twitter/X thread with a real screenshot, not just text

## Ongoing (weekly, first month)
- [ ] Respond to every issue within 48h, even if just "triaging, thanks"
- [ ] One awesome-list PR opened
- [ ] Star count and external-PR count logged (see analytics cell below)
"""

(PROJECT_ROOT / "LAUNCH_CHECKLIST.md").write_text(LAUNCH_CHECKLIST_MD)
print(f"Wrote {PROJECT_ROOT / 'LAUNCH_CHECKLIST.md'}")

# Star-history badge — free, no API key, embeds a live chart in the README once the repo has stars
star_history_badge = (
    "[![Star History Chart](https://api.star-history.com/svg?repos=gouravkhanijoe/paper-distiller&type=Date)]"
    "(https://star-history.com/#gouravkhanijoe/paper-distiller&Date)"
)
print("\nAdd to README once repo is public:\n", star_history_badge)
````

## The Metric That Actually Matters: External Contribution Rate

Stars are a vanity metric — they correlate weakly with usage and not at all with project health. The metric worth tracking weekly is the **external contribution rate**: what fraction of issues/PRs come from people who aren't you, and how fast do you respond to them.

A repo with 500 stars and zero external PRs after 3 months is not a healthy OSS project — it's a popular gist. A repo with 20 stars and 3 external contributors who each came back a second time is the one that compounds.


````
from dataclasses import dataclass
from datetime import date

@dataclass
class Activity:
    date: date
    author: str
    kind: str          # "issue" | "pr"
    is_maintainer: bool
    first_response_hours: float | None = None

# Mock activity log — swap for a real `gh api repos/:owner/:repo/issues` pull once the repo is live
mock_log = [
    Activity(date(2026, 7, 3), "gouravkhanijoe", "issue", True),
    Activity(date(2026, 7, 4), "gouravkhanijoe", "issue", True),
    Activity(date(2026, 7, 6), "alice_dev", "issue", False, first_response_hours=6.0),
    Activity(date(2026, 7, 8), "bob_oss", "pr", False, first_response_hours=18.0),
    Activity(date(2026, 7, 12), "alice_dev", "pr", False, first_response_hours=4.0),   # alice came back!
    Activity(date(2026, 7, 15), "carol_ml", "issue", False, first_response_hours=30.0),
]

def contribution_report(log: list[Activity]) -> dict:
    external = [a for a in log if not a.is_maintainer]
    external_authors = {a.author for a in external}
    repeat_authors = {a for a in external_authors
                       if sum(1 for x in external if x.author == a) > 1}
    response_times = [a.first_response_hours for a in external if a.first_response_hours is not None]
    return {
        "total_activity": len(log),
        "external_activity": len(external),
        "external_contribution_rate": round(len(external) / len(log), 2) if log else 0.0,
        "unique_external_contributors": len(external_authors),
        "repeat_contributors": len(repeat_authors),
        "avg_first_response_hours": round(sum(response_times) / len(response_times), 1) if response_times else None,
        "sla_48h_met_pct": round(100 * sum(1 for t in response_times if t <= 48) / len(response_times), 0) if response_times else None,
    }

report = contribution_report(mock_log)
for k, v in report.items():
    print(f"{k:32s}: {v}")

# 💡 EXPERIMENT: once paper-distiller is on GitHub, replace `mock_log` with real data via
# `gh issue list --json author,createdAt,isPullRequest --repo gouravkhanijoe/paper-distiller`
# and re-run this weekly. Track external_contribution_rate over time — that's your real growth curve.
````

## 10 Pitfalls in OSS Growth

| # | Pitfall | Why it hurts |
|---|---------|---------------|
| 1 | Launching before the quickstart is verified end-to-end | First impressions during a launch spike don't get a second chance |
| 2 | A green CI badge on a workflow that doesn't actually fail on test failure | Trust violation the moment a contributor checks — worse than no badge |
| 3 | Zero labeled issues at launch | Repo looks abandoned or "not ready for contributors," even if the code is solid |
| 4 | Writing a custom Code of Conduct instead of Contributor Covenant | Reviewers/companies recognize and trust the standard one instantly; custom ones raise questions |
| 5 | Stacking Show HN + Reddit + Twitter on the same day | Traffic surge overwhelms your ability to respond to comments/issues in real time |
| 6 | CONTRIBUTING.md that doesn't include the exact commands to run locally | Ambiguity is silent drop-off — people don't ask, they leave |
| 7 | Treating stars as the success metric | Stars don't correlate with usage or health; external contribution rate does |
| 8 | No response to first external issue for days | The single highest-leverage moment to retain a contributor is the first 48 hours |
| 9 | Badges pointing to the wrong package/repo name after a rename | Broken badges (404) look worse than no badges |
